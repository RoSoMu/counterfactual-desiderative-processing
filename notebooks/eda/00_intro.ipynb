### 1. Research question & linguistic problem

This project investigates how **L1-Spanish adult learners of English process and extend past-subjunctive meaning in L2 English**.

Spanish and English differ in how they encode subjunctive mood. In English third conditionals, the antecedent uses the **Past Perfect** to refer to a counterfactual past:

> *If I had taken a different route, I would have arrived earlier.*

For an L1-Spanish learner, this creates a cross-linguistic processing problem: a tense form that is morphologically indicative in English is used in a context corresponding to past-subjunctive meaning.

The experiment asked whether the **counterfactual context surrounding this form** could make that L2 mapping easier to process and subsequently extend to a different construction requiring the same tense:

> *I wish I had taken a different route.*

#### Research question

**Can properties of a third-conditional counterfactual context facilitate L1-Spanish learners' subsequent use of the Past Perfect with past-subjunctive meaning in an English desiderative sentence?**

### 2. Suggested theoretical readings

The experiment draws primarily on three theoretical contributions from sentence processing and counterfactual-thinking research. The brief summaries below provide the concepts needed to understand the experimental manipulations; the original publications are listed for readers who want the full theoretical account.

#### Gibson, 1998

_Syntactic Prediction Locality Theory_ (SPLT) connects sentence-processing complexity with available computational resources and distinguishes memory cost associated with maintaining obligatory syntactic requirements from integration cost incurred when incoming material must be integrated with earlier structure. Both are strongly affected by locality: keeping a prediction unresolved longer increases memory cost, while greater distance between dependent elements increases integration cost.

#### Roese & Epstude, 2017

The _functional theory of counterfactual thinking_ connects episodic counterfactual thinking to goals and behaviour regulation—counterfactuals often represent alternative pathways by which failed goals might have been achieved and can inform future action. It explicitly describes how counterfactual thought arises systematically in response to goal blockage, rather than randomly.

#### Girotto et al., 2007

The information available to the thinker depends on their role. Across their actor/reader studies, readers tended to undo the protagonist’s choice, whereas people who had actually experienced the unsuccessful problem-solving episode tended to alter features of the problem instead. Their account emphasizes _differential availability/salience of information to actors and readers_.

### 3. Experimental logic

The experiment used **third-conditional bacground sentences as contexts** to test whether different properties of a counterfactual situation affected learners' subsequent production of a desiderative sentence.

Each experimental item contained two stages:

1. **Counterfactual context** — participants encountered a third conditional containing the Past Perfect in its antecedent and *would have + past participle* in its consequent.

2. **Desiderative response** — participants were then presented with a wish about the same past situation in a fill-the-gap multiple-choice format. The desiderative sentence was provided except for the target verb form, for which four alternatives were available: one systematic/target response using the Past Perfect, one intuitive/L1-transfer response, and two L2-other alternatives. This created an opportunity to select and reuse the Past Perfect with past-subjunctive meaning without requiring participants to construct the desiderative sentence themselves.

The counterfactual contexts varied along three theoretically motivated dimensions:

- **Goal** — whether the situation involved a frequent goal, a non-frequent goal, or no explicit goal.
- **Agency** — whether the antecedent contained an explicit animate agent or expressed the event without explicit agency.
- **Focus** — whether the counterfactual was framed around the participant (`I`) or another person (`they`).

The original experimental logic was that these properties could alter the cognitive conditions under which the third conditional was processed. Contexts expected to require fewer processing resources and/or provide a stronger functional basis for counterfactual reasoning were predicted to facilitate subsequent target use of the Past Perfect in the desiderative.

The following sections present each manipulation separately, using the original experimental items to show what participants actually encountered and the theoretical reasoning behind each prediction.

#### Why was Goal expected to matter?

The Goal manipulation combined two complementary ideas: the **processing demands of the linguistic input** and the **functional role of counterfactual thinking**.

##### 🧠 BRAIN — Processing economy

From a processing perspective, the experiment assumed that more **frequent, familiar and predictable event representations** would require fewer computational resources to process. Following Gibson's resource-based account of sentence processing, reducing demands elsewhere in the sentence was expected to leave more resources available for processing the difficult L2 form.

he frequent-goal items represented ordinary, readily accessible relationships between actions and outcomes (*save money → go on holiday; practise sports → lose weight*).

The non-frequent-goal items remained causally coherent, but represented less routine and less predictable situations. They also introduced an **external referent** through *my family / my family's* (*create an online platform → save my family's business; invest in technology → secure my family's wealth*), adding information that had to be represented alongside the self-focused agent and the counterfactual event and, potentially, adding cost.

The no-goal items differed more fundamentally: instead of representing an agent performing a goal-directed action, they presented an external event or circumstance, with the person in the consequent functioning as the experiencer of an internal state (feel relaxed; feel invincible).

##### 🧠 MIND — Functional counterfactuality

From the functional perspective of counterfactual thinking, counterfactuals are closely connected to **goals, goal blockage and future behaviour**. Imagining how a negative outcome could have been different can identify alternative means of achieving a goal and provide information relevant to future action.

The frequent-goal conditions were therefore expected to provide the strongest functional context: their goals represented familiar actions and outcomes that could plausibly recur and be acted upon in the future. Non-frequent goals also provided a meaningful goal structure, but concerned less routine and less readily repeatable situations.

The non-frequent-goal items involved **less routine and more emotionally consequential goals**, centred on the welfare or financial security of the participant's family. Although strongly goal-directed, these situations were less readily repeatable and potentially more emotionally demanding.

The no-goal items provided the weakest functional structure for this purpose. Although their consequents described desirable internal states (*feeling relaxed* or *feeling invincible*), they were not constructed around an explicit blocked goal whose alternative means could guide subsequent goal-directed behaviour.

#### Original hypothesis

Frequent-goal counterfactuals were expected to provide the most favourable combination of **processing accessibility** and **functional relevance**, leaving more resources available for the difficult L2 form while providing a meaningful counterfactual framework in which that form occurred.

Target Past Perfect selection in the subsequent desiderative was expected to follow the hierarchy:

### 5. Agency Manipulation → rationale → hypothesis

```python
# Filter by Focus, and drop duplicates

agent_strings = counterfactuals[
    (counterfactuals["Focus"] == "I")      
]

agent_strings = agent_strings[
    ["Agent", "Goal_Type", "Item", "Item_string"]
].drop_duplicates()

# Sort Agent, Agent = 1 first

agent_order = [1, 0]

agent_strings["Agent"] = pd.Categorical(
    agent_strings["Agent"],
    categories=(agent_order),
    ordered=True
    )

# Sort goal type, frequent goal first

goal_order = ["goal_frequent", "goal_non_frequent", "no_goal"]

agent_strings["Goal_Type"] = pd.Categorical(
    agent_strings["Goal_Type"],
    categories=goal_order,
    ordered=True
    )

agent_strings = agent_strings.sort_values(
    ["Agent", "Goal_Type", "Item"]
    ).reset_index(drop=True)

pd.set_option("display.max_colwidth", None)

```

```python
agent_display = agent_strings.pivot(
    index=["Goal_Type", "Item"],
    columns="Agent",
    values="Item_string"
)
agent_display = agent_display.rename(
    columns={
        1: "Explicit Agent (Agent = 1)",
        0: "Implicit Agency (Agent = 0)"
        }
).reset_index()

agent_display
```

#### Why was Agency expected to matter?

The Agency manipulation tested whether explicitly representing the participant as the **animate agent of the counterfactual action** would facilitate subsequent target selection.

Across matched items, the consequent was kept constant while the antecedent changed. In the explicit-agency condition (`Agent = 1`), the participant was encoded as the grammatical subject and agent of the action (*I had saved...*, *I had invested...*, *I had climbed...*). In the implicit-agency condition (`Agent = 0`), the same counterfactual situation was expressed without overtly representing the participant as its agent (*savings had been sufficient...*, *technological investment had been encouraged...*, *there had been a rainstorm...*).

As with Goal Type, the original prediction combined a **processing** rationale with a **functional counterfactual** rationale.

##### 🧠 BRAIN — Processing economy

From a processing perspective, an **animate, explicit agent performing an action** was expected to provide a relatively accessible mapping between the sentence's syntactic structure and its thematic roles: an agent performs an action that contributes to an outcome.

The `Agent = 0` items removed this overt agent-action relationship. Depending on the item, the antecedent instead used an inanimate subject, a passive construction, an existential construction, or a state. These alternatives were expected to make the causal and thematic structure of the antecedent less immediately accessible.

Within the original resource-based account, greater accessibility of the agent-action relationship was expected to reduce processing demands elsewhere in the counterfactual, leaving more resources available for processing the difficult L2 tense form.

##### 🧠 MIND — Functional counterfactuality

From a functional perspective, counterfactual thinking often identifies **how an unsuccessful outcome might have been changed** and can make alternative means of achieving a goal available for future behaviour.

Explicit agency made the participant's own action directly available as a potentially **controllable and mutable element** of the counterfactual: *I could have saved, practised, created, invested, heard,* or *climbed*. The antecedent therefore provided an explicit link between an action attributed to the self and the alternative outcome that followed from it.

When agency was implicit, the alternative situation could still support the same outcome, but the participant was not overtly represented as the person performing the action that brought it about.

#### Original hypothesis

**Counterfactuals with explicit agency (`Agent = 1`) were expected to facilitate target Past Perfect selection more than their implicit-agency counterparts (`Agent = 0`).**

**Explicit agency > Implicit agency**

The explicit-agent condition was therefore expected to provide the more favourable combination of **accessible linguistic structure** and **functionally relevant causal control**.

## 6. Focus manipulation

### Why was Focus expected to matter?

The Focus manipulation tested whether framing the counterfactual around the **self (`I`) rather than another person (`they`)** would facilitate subsequent target selection.

Unlike Goal Type and Agency, Focus was manipulated **between participants**: one group encountered self-focused counterfactuals, while the other encountered corresponding other-focused versions. Across matched items, the underlying event and outcome were preserved while the person whose alternative past was represented changed.

The original prediction again combined a **processing** rationale with a **functional counterfactual** rationale, while drawing on evidence that the information available during counterfactual reasoning may differ depending on an individual's relationship to the event.

#### 🧠 BRAIN — Processing and accessibility

From a processing perspective, self-focused counterfactuals were expected to provide a more immediately accessible representation than corresponding counterfactuals centred on another person.

The self-focused condition maintained the participant as the central discourse referent (`I`), whereas the other-focused condition introduced and maintained another individual (`they`). Within the original resource-based account, this additional referential representation was expected to increase processing demands, leaving comparatively fewer resources available for processing the difficult L2 form.

The predicted advantage was therefore not based simply on the pronoun `I` being linguistically easier than `they`, but on the assumed accessibility of a counterfactual representation organised around the self rather than an external referent.

#### 🧠 MIND — Functional counterfactuality

The functional theory provides a complementary reason to expect a self-focus advantage. Roese and Epstude describe spontaneous counterfactual thought as typically **personal and self-focused**: modifying one's own behaviour can provide information that is directly relevant to subsequent goal pursuit.

Girotto et al. add an important distinction. In their experiments, people who had **actually experienced an unsuccessful decision** generated different counterfactual alternatives from people who merely **read about another person's experience**. Actors tended to modify features or means involved in reaching the unsuccessful outcome, whereas readers more often undid the protagonist's choice. The authors attributed this difference to the information made available by the individual's role and experience.

The present experiment did **not** reproduce that actor–reader distinction. Participants in both Focus conditions completed an offline language task and had not actually performed or experienced the actions described in the counterfactuals. Operationally, therefore, `I` and `they` represent **self-focus and other-focus**, not genuine actor and reader roles.

Nevertheless, the original study asked whether linguistically placing the participant inside the counterfactual (`I`) might provide some of the accessibility and functional advantages associated with self-focused counterfactual thought, compared with representing the same alternative past around another person (`they`).

### Original hypothesis

**Self-focused counterfactuals (`I`) were expected to facilitate target Past Perfect selection more than corresponding other-focused counterfactuals (`they`).**

**Self-focus (`I`) > Other-focus (`They`)**

Self-focus was therefore expected to provide the more favourable combination of **referential accessibility** and **functional relevance to the self**.

### 7. Original predicted extremes

The three manipulations produced directional hypotheses about Goal Type, Agency and Focus. However, the original framework did **not** specify a complete ranking of all possible combinations.

In particular, the theoretical properties manipulated by the experiment were not assumed to have equivalent weights. A configuration combining favourable and less favourable properties could therefore not be placed reliably above or below another mixed configuration.

The original predictions did, however, identify two theoretical extremes.

#### Most favourable predicted configuration

**Frequent goal + Explicit agency + Self-focus (`I`)**

This configuration combined a frequent and repeatable goal, an explicit animate agent performing a controllable action, and a counterfactual centred on the self.

It was therefore expected to provide the strongest combination of **processing accessibility** and **functional counterfactual relevance**, and consequently the most favourable context for subsequent target Past Perfect selection.

> *If I had saved money, I would have gone on holidays.*

> *If I had practised sports, I would have lost weight.*

#### Least favourable predicted configuration

**No goal + Implicit agency + Other-focus (`They`)**

This configuration lacked an explicit blocked-goal structure and an overt animate agent in the antecedent, while representing the counterfactual around another person.

The no-goal condition also differed in its **event and thematic-role structure**. Rather than representing an agent performing a goal-directed action, the antecedent introduced an external event or circumstance (*a rainstorm; a mountain cliff*), while the person represented in the consequent became the **experiencer of an internal state** expressed through *feel* (*they would have felt relaxed; they would have felt invincible*).

Under the original framework, this configuration was therefore expected to provide the weakest combination of the processing and functional properties hypothesised to facilitate subsequent target selection.

> *If there had been a rainstorm, they would have felt relaxed.*

> *If there had been a mountain cliff, they would have felt invincible.*

These configurations represent the **predicted endpoints**, not a complete ranking of the intermediate experimental conditions.
