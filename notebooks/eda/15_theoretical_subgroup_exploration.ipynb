# Consequent Reversibility Exploration

This and following notebooks explore **consequent reversibility** as a potential categorical variable in order to look into further explaining findings the following findings:
- self-focus asymmetries between goal-contexts
- transfer domination 
- no-goal context different behaviour
- reader-agent effects

Consequent reversibility would also match emotional load inasmuch as the goal/ items with consequent_reversibility = Yes would also be the one emotionally loaded.

1. Overall Subgroup
2. Actors
3. Readers

Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Read data and sanity checks

```python
subgroup_theoretical = pd.read_csv("../../data/processed/subgroup_theoretical.csv")
```

```python
subgroup_theoretical[
    [
        "Goal_Type",
        "Item",
        "Consequent_Reversibility"
    ]
].drop_duplicates()
```

### 1. Overall Subgroup

Response Proportion by Consequent Reversiblity

```python
reversibility_props= pd.crosstab(
    subgroup_theoretical["Consequent_Reversibility"],
    subgroup_theoretical["Response_Full"],
    normalize="index"
)

reversibility_props
```

```python
reversibility_props.plot(kind = "bar", stacked=True)

plt.title("Response Proportions Consequent Reversibility")
plt.legend(title="Response", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.xlabel("Consequent Reversibilty")
plt.ylabel("Proportion")
plt.show()
```

Response Counts by Consequent Reversiblity

```python
reversibility_counts= pd.crosstab(
    subgroup_theoretical["Consequent_Reversibility"],
    subgroup_theoretical["Response_Full"]
)

reversibility_counts
```

For irreversible consequence ( = no), as expected, proportion of transfer responses in higher, there are fewer L2 alternative exploration and fewer correct responses. All 3 move in the expected direction and not only transfer.

Response Proportion by Consequent Reversiblity for Readers

```python
reversibility_props_r= pd.crosstab(
    subgroup_theoretical_readers["Consequent_Reversibility"],
    subgroup_theoretical_readers["Response_Full"],
    normalize="index"
)

reversibility_props_r
```
