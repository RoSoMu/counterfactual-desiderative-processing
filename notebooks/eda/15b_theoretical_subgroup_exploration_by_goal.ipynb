Import libraries

Read data and sanity checks

### 1. Overall Subgroup

Response Proportion by Goal Type and Consequent Reversiblity

```python
goal_reversibility_props= pd.crosstab(
    [subgroup_theoretical["Goal_Type"], subgroup_theoretical["Consequent_Reversibility"]],
    subgroup_theoretical["Response_Full"],
    normalize="index"
)

goal_reversibility_props
```

```python
goal_reversibility_props.plot(kind = "bar", stacked=True)

plt.title("Response Proportions by Goal Type and Consequent Reversibility")
plt.legend(title="Response", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.xlabel("Consequent Reversibilty")
plt.ylabel("Proportion")
plt.show()
```

Response Counts by Goal Type and Consequent Reversiblity

```python
goal_reversibility_counts= pd.crosstab(
    [subgroup_theoretical["Goal_Type"], subgroup_theoretical["Consequent_Reversibility"]],
    subgroup_theoretical["Response_Full"]
)

goal_reversibility_counts
```

- Goal frequent is the best perfroming category and explores more L2-alternatives.
- Goal non frequent is the worst performing category, the highest in transfer and reduces the most L2-alternatives. 
- No-goal contexts sit in between.

### 2. Actors

Actors subgroup proportions

```python
subgroup_theoretical_actors.shape
```

Response Proportion by Goal Type and Consequent Reversiblity for Actors

```python
goal_reversibility_props_a= pd.crosstab(
    [subgroup_theoretical_actors["Goal_Type"],subgroup_theoretical_actors["Consequent_Reversibility"]],
    subgroup_theoretical_actors["Response_Full"],
    normalize="index"
)

goal_reversibility_props_a
```

```python
goal_reversibility_props_a.plot(kind = "bar", stacked=True)

plt.title("Response Proportions by Goal Type Consequent Reversibility for Actors")
plt.legend(title="Response", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.xlabel("Consequent Reversibilty")
plt.ylabel("Proportion")
plt.show()
```

Response Counts by Goal Type and Consequent Reversiblity for Actors

```python
goal_reversibility_counts_a= pd.crosstab(
    [subgroup_theoretical_actors["Goal_Type"],subgroup_theoretical_actors["Consequent_Reversibility"]],
    subgroup_theoretical_actors["Response_Full"]
)

goal_reversibility_counts_a
```

- For goal frequent, proportion of correct responses is 41%, the highest, and proportion of L1-transfer, proportion is 33%, the lowest.
- For goal non frequent, proportion of correct responses is still reasonably high, 33% and, for no-goal contexts, 29%, the lowest. If the emotional load (irreversibility) was responsible ofr this, I would expect non frequent goal to collapse below no goal because it carries liability; but it doesn't.
- This, however, seems to signal that no-goal contexts could behave differently: there seem to be no objective and no behavioural strategy, just an experience. 

### 3. Readers

Readers subgroup proportions

```python
subgroup_theoretical_readers = subgroup_theoretical[subgroup_theoretical["Focus"] == "They"].copy()
```

```python
subgroup_theoretical_readers.shape
```

Response Proportion by Goal Type and Consequent Reversiblity for Readers

```python
goal_reversibility_props_r= pd.crosstab(
    [subgroup_theoretical_readers["Goal_Type"],subgroup_theoretical_readers["Consequent_Reversibility"]],
    subgroup_theoretical_readers["Response_Full"],
    normalize="index"
)

goal_reversibility_props_r
```

```python
goal_reversibility_props_r.plot(kind = "bar", stacked=True)

plt.title("Response Proportions by Goal Type and Consequent Reversibility for Readers")
plt.legend(title="Response", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.xlabel("Consequent Reversibilty")
plt.ylabel("Proportion")
plt.show()
```

Response Counts by Goal Type and Consequent Reversiblity for Readers

```python
goal_reversibility_counts_r= pd.crosstab(
    [subgroup_theoretical_readers["Goal_Type"],subgroup_theoretical_readers["Consequent_Reversibility"]],
    subgroup_theoretical_readers["Response_Full"]
)

goal_reversibility_counts_r
```

- Frequent goal and no-goal context have identical proportions of L1-transfer (46%) and very close correct responses (31% vs. 27%). Slighther higher L2 exploration for no-goal contexts. That's not even close to what actors did, which showed higher L1-transfer for no-goal contexts (58% vs. 33% for frequent goals), lower correct responses (29% vs. 41% for frequent goals) and far lower L2 exploration (13% vs 26% for frequent goals). This is an interesting assymetry.

- The increase of L1 transfer responses for non-frequent goals in readers (73%) and the lack of L2-alternative exploration (4%) is also very different from what we say in actors. 

- It could suggest that purpose may be a procesing resource for actors whereas distance may be a processing resource for readers; what could be facilitating for one could be irrelevant for the other.
