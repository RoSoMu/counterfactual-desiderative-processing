### 1. Overall Subgroup

Response Proportion by Agent and Consequent Reversiblity

```python
agent_reversibility_props= pd.crosstab(
    [subgroup_theoretical["Agent"], subgroup_theoretical["Consequent_Reversibility"]],
    subgroup_theoretical["Response_Full"],
    normalize="index"
)

agent_reversibility_props
```

```python
agent_reversibility_props.plot(kind = "bar", stacked=True)

plt.title("Response Proportions by Agent and Consequent Reversibility")
plt.legend(title="Response", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.xlabel("Consequent Reversibilty")
plt.ylabel("Proportion")
plt.show()
```

- When the agent is implicit, differences in response proportions is rather small showing how consequent reversibility could help exploration by decreasing transfer slightly but, in fact, in terms of correctness, they are almost identical.

- When the agent is explicit, reversibility matters more. Consequent reversibility reduces facilitates restructure showing an increase in correct responses, L1-transfer lowers and L2-other responses also increase. Consequent reversibilty seem to matter so long as someone is attached to the outcome.
