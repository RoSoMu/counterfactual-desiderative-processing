1. Overal Subgroup

Summary Table: responses proportions

```python
response_props = pd.crosstab(
    [subgroup_theoretical["Goal_Type"], subgroup_theoretical["Agent"], subgroup_theoretical["Consequent_Reversibility"]],
    subgroup_theoretical["Response_Full"],
    normalize="index"
)

response_props
```

Summary Table: responses count

```python
response_count = pd.crosstab(
    [subgroup_theoretical["Goal_Type"], subgroup_theoretical["Agent"], subgroup_theoretical["Consequent_Reversibility"]],
    subgroup_theoretical["Response_Full"]
)

response_count
```

- The apparent differences do not seem to be driven by tiny cells. 
- The differences L2 alternative cells once agent is introduced (goal frequent 14 --> 2; goal non frequent 9 ---> 1) are substantial.
- The no goal condition really seem to behave differently: agency doesn't seem to harm performance; if anything, it redistributes transfer while preserving conrrect responses. 

In summary:
- The introduction of an agent appears to suppress L2 responses across all goal conditions. This effect is particularly pronounced in the goal_non_frequent condition, where L1 transfer becomes dominant and correct responses decrease.

- The actor group is more structured. 
- Reversibility has surprisingly little effect when agency is absent, but becomes highly consequential when agency is explicit.
