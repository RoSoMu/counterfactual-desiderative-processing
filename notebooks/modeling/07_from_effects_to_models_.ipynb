# Counterfactual-Desiderative Processing: From Effects to Models

## 1. Purpose of this notebook
This notebook explains the logic behind the modelling notebooks that follow. It documents the analytical decisions made after the exploratory, the effect analysis and response opportunity phases and provides a map for readers.

The purpose of this project is not to identify isolated statistically significant predictors, but to investigate whether behavioural relationships remain coherent within a theoretically constrained and comparatively homogeneous subset of the data. 

After having explored and analysed the effect on the combined and split datasets separately, it seems that the no-goal observations are simply not constrained by the same cognitive architecture as the goal observations. Without the goal manipulations, there is less to organize their behaviour into a common pattern; and heterogeneous stategies would be difficult to capture by logistic modeling.

## 2. Main research pathway
There are 2 different stages:

- Stage 1: Escape from L2  
- Stage 2: Resolution after escape (resolving by L1-transfer vs. resolving correctly)


## 3. Subgroup
Originally, participants were selected according to perceived proficiency (B2.1 to C1) and professional use criteria.
Participant restriction: 336 --> 198 observations.

## 4. Why escape and resolution are modelled separately
- Escape_L2: it captures whether participants move away from L2_other options.  
- Resolution: it captures what they do after escaping L2_other options: correct response vs L1_transfer.

## 5. Why Consequent_Reversibility was dropped
Consequent_Reversibility was explored alongside the other orginal predictors (agent and goal type), but it did not provide a clearer theoretical or statistical contribution. It also overlapped conceptually with the goal type structure, so it was removed from the main modelling pathway (Notebooks 15, 15b, 15c, 15d and 16)
