# L2 Escape Models

The following models take into account that, throughout analyses, *no_goal* seems to behave entirely differently to the other two Goal Type conditions (*goal_frequent* and *goal_non_frequent*). 

Previous focus structure effects (see notebooks 11 to 6) as well as furhter exhaustive modelling (see notebooks 11_appendix vs. 12_appendix) show consistently the heterogeneity of combined data; by spliting data into goal contexts and no-goal contexts, both effects as well as models improved despite the pruning of observations. No-goal contexts are also analysed separately; however, and although results are coherent with what is seen by comparing combined data to goal behaviour data, the sample is clearly not enough to show anything significant from no-goal contexts.

Taking into account the effects summary (notebook 5) and response opportunity analysis (notebook 6), these were chosen to be the best models to predict participants escaping L2-other options.

3. Logistic Models

Import Libraries

```python
import statsmodels.formula.api as smf
import statsmodels.api as sm
from scipy.stats import chi2
```

Sanity checks:

```python
goals["escape_L2"].value_counts()
```

```python
no_goal["escape_L2"].value_counts()
```

**Goal Contexts**

**Main Effect Models**

**E02**

escape_L2 ~ Agent

```python
E02 = smf.logit(
    "escape_L2 ~ Agent",
    data=goals
    ).fit()

print(E02.summary())
```

When the data was combined, agency on its own's p-value = 0.013 (coeff = 1.1086). Once the data is the goal split is introduced, agency coefficient increases (1.7327) and p-value improves ( 0.007). Thus, after removing 66 observations, the effect gets actually stronger. 

**Additive Models**

**E04**

escape_L2 ~ Focus + Agent

```python
E04 = smf.logit(
    "escape_L2 ~  Focus + Agent",
    data=goals
    ).fit()

print(E04.summary())
```

Focus on its own couldn't predict anything. Again, once it is added to the model with agency, while it is still non-significant (p-value 0 0.080), it is much more informative than boefore. The coefficient also grows from 0.694 to 0.876; so, once agency is accounted for, focus explains additional variation in the data.

The whole model's LLR p-value (0.00135) is also better than agent alone (0.00163); not a lot but not requiring an interaction.

This model explains a lot of variation and demostrates parsimony.


**E06**

escape_L2 ~ Agent + Goal_Type

```python
E06= smf.logit(
    "escape_L2 ~  Agent + Goal_Type",
    data=goals
    ).fit()

print(E06.summary())
```

This is a good model even though the goal_non_frequent's p is still not siginificant. It is contributing to the model. Given this, a full additive model that was not considered in the beginning should be looked into.

**E07**

escape_L2 ~ Focus + Agent + Goal_Type

```python
E07= smf.logit(
    "escape_L2 ~  Focus + Agent + Goal_Type",
    data=goals
    ).fit()

print(E07.summary())
```

Goal Type's p-value improves a little bit and the coefficient increases. The whole model is better (LLR p-value: 0.001709 vs. 0.002758) although not better than the focus-agency additive model's p (0.001351). 

Yet, this model shows a better picture of how counterfactual processing is shaped by considering the 3 cognitive dimensions (Focus, Agency and Goal_Type). In this sense, Agency shows a strong contribution, Focus shows a consistent positive contribution approaching significance and Goal Type shows a modest positive contribution. 

**No-goal Contexts**

**Main Effect Models**

**E12**

escape_L2 ~ Focus

```python
E12 = smf.logit(
    "escape_L2 ~ Focus",
    data=no_goal
    ).fit()

print(E12.summary())
```

**E13**

escape_L2 ~ Agent

```python
E13 = smf.logit(
    "escape_L2 ~ Agent",
    data=no_goal
    ).fit()

print(E13.summary())
```

- While goal participants seem to produce a coherent behaviourla structure, the no_goal participants don't. They don't seem to have the same behaviour constraints. 

- In addition, and even though there are only 66 observations to render anything statistically conclusive, escaping the L2 landscape for no-goal context modelling shows to be coherent with EDA; if anything helped participants escape the L2 landscape, it would be more likely to be Focus (p-value = 0.146) and surely not Agency (p-value = 0.809) -quite the opposite to what happens when looking into goal-context behaviour.

4. Model Comparisons

```python
from scipy.stats import chi2

def lr_test(model_small, model_large):

    LR = 2 * (model_large.llf - model_small.llf)

    df = model_large.df_model - model_small.df_model

    p = chi2.sf(LR, df)

    return pd.Series({
        "LR": LR,
        "df": df,
        "p_value": p
    })
```

**Unique contributions to additive models**

Agent (E02)

↓

Focus + Agent (E04)

```python
lr_test(E02, E04)
```

- Does Focus add information after Agent?

This is week evidence that, once the variability added by Agent is removed, Focus contributes something to escaping the L2. Not to claim it statistically, but enough to have it in the additive model.

Agent (E02)

↓

Agent + Goal_Type (E06)

```python
lr_test(E02, E06)
```

- Does Goal_Type add information after Agent?

No evidence that goal type adds any information once the variability added by Agent is removed. 

**Unique contributions to the full additive model**

Focus + Agent (E04)

↓

Focus + Agent + Goal_Type (E07)

```python
lr_test(E04, E07)
```

- Does Goal_Type add information after Focus and Agent?

No. Once Focus and Agency are already in the Model, Goal_Type doesn't contribute enough additional information. This confirms that leaving the L2-other alternatives is primarily driven by Agency. 

Agent + Goal_Type (E06)

↓

Focus + Agent + Goal_Type (E07)

```python
lr_test(E06, E07)
```

- Does Focus add information after Agent and Goal_Type?

This is week evidence that, once the variability added by no_goals is removed, Focus contributes something to escaping the L2. Not to claim it statistically, but enough to have it in the additive model.

5. Summary

- Agency predicts escaping alternative L2 responses among participants who actually have a goal to protect. Thus, we could ask if ,once there isn't a goal scenario, what could agency help or contribute to. 

- Goal Type consistently exhibited positive coefficients across additive models, although its independent contribution remained modest and did not reach conventional levels of statistical significance.

- Given results of the full additive model, escape responses are best understood as the joint product of agency, perspective, and goal structure, although agency provides the strongest independent contribution. Adding Goal Type preserves the pattern of effects and yields a coherent cognitive account in which agency remains the dominant predictor while focus and goal structure contribute more modestly.

From Model Comparisons:

- There is week eveidence that Focus contributes something to escaping the L2. In contrast, Goal_Type does not significantly improve a model already containing Focus and Agency.

- Separating the no-goal condition reduced heterogeneity in the remaining data, resulting in clearer predictor hierarchies and more interpretable model comparisons.

- Regarding the no-goal subgroup, model comparisons suggested that Focus captured more unique inforamtion than Agency. Whereas the oppostite hierarchy characterised the goal conditions. They are underpowered because of sample size, but the pattern is consistent with the descriptive anaylyses.
