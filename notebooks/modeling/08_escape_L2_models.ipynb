# L2 Escape Models

The following models take into account that, throughout analyses, *no_goal* seems to behave entirely differently to the other two Goal Type conditions (*goal_frequent* and *goal_non_frequent*). 

Previous focus structure effects (see notebooks 11 to 6) as well as furhter exhaustive modelling (see notebooks 11_appendix vs. 12_appendix) show consistently the heterogeneity of combined data; by spliting data into goal contexts and no-goal contexts, both effects as well as models improved despite the pruning of observations. No-goal contexts are also analysed separately ((notebook 11)); however, and although results are coherent with what is seen by comparing combined data to goal behaviour data, the sample is clearly not enough to show anything significant from no-goal contexts.

Taking into account the effects summary (notebook 5) and response opportunity analysis (notebook 6), starting from the main effect identified in the previous analyses, predictors were added only when supported by opportunity.

Read Data

```python
subgroup_theoretical = pd.read_csv("../../data/processed/subgroup_theoretical.csv")
```

1. Create escape_2

Binary Outcome: responses that are not "L2_other": those who managed to escape L2_other by choosing either a *correct* response or an *L1_transfer* response.

```python
subgroup_theoretical["escape_L2"] = (
    subgroup_theoretical["Response_Full"] != "L2_other"
).astype(int)
```

Split data into goals vs. no_goals

```python
goals = subgroup_theoretical[subgroup_theoretical["Goal_Type"]!= "no_goal"].copy()
```

```python
no_goal = subgroup_theoretical[subgroup_theoretical["Goal_Type"]== "no_goal"].copy()
```

2. Counts and Proportions for goals and no_goal.

Escape_L2 responses count for goals:

```python
goal_escape_L2_counts = pd.crosstab(
    goals["escape_L2"],
    goals["Response_Full"],
    margins= True)

goal_escape_L2_counts
```

Escape_L2 responses proportions for goals:

```python
goals_escape_L2_props = goals["escape_L2"].value_counts(normalize = True)
goals_escape_L2_props
```

From 132 observations in the goal group, 80.3% of responses managed to escape L2_other options. 

Counts and Proportions by Condition:

```python
goals_counts_condition = pd.crosstab(
    [goals["Goal_Type"], goals["Agent"], goals["Focus"]],
    goals["escape_L2"],
    margins = True)

goals_counts_condition
```

```python
goal_props_condition = pd.crosstab(
    [goals["Goal_Type"], goals["Agent"], goals["Focus"]],
    goals["escape_L2"],
    normalize= "index")

goal_props_condition
```

Within goals, Agency seems to assist in the process of escaping the L2_other options.

- As previously mentioned, if there is a goal representation available, agency seems to help participants commit by leaving the L2 space.

3. Logistic Models

Import Libraries

```python
import statsmodels.formula.api as smf
import statsmodels.api as sm
from scipy.stats import chi2
```

Sanity checks:

```python
goals["escape_L2"].value_counts()
```

```python
no_goal["escape_L2"].value_counts()
```

**Goal Contexts**

**Main Effect Models**

**E02**

escape_L2 ~ Agent

```python
E02 = smf.logit(
    "escape_L2 ~ Agent",
    data=goals
    ).fit()

print(E02.summary())
```

Agency accounts for most of the explainable variation. 

**Additive Models**

**E04**

escape_L2 ~ Focus + Agent

```python
E04 = smf.logit(
    "escape_L2 ~  Focus + Agent",
    data=goals
    ).fit()

print(E04.summary())
```

Once Focus is added to the model with Agency, while it is still non-significant (p-value 0 0.080), it is much more informative than before. Focus explains additional variation in the data.

The whole model's LLR p-value (0.00135) is also better than agent alone (0.00163); not a lot but not requiring an interaction.

This model explains a lot of variation and demostrates parsimony.In addition, it shows coherent with directional effects across conditions from the opportunity analysis.

**E06**

escape_L2 ~ Agent + Goal_Type

```python
E06= smf.logit(
    "escape_L2 ~  Agent + Goal_Type",
    data=goals
    ).fit()

print(E06.summary())
```

This is a good model even though the goal_non_frequent's p is still not siginificant, it captures another descriptive pattern. 

Non-frequent goals tend to suppress compteing L2 strategies more strongly. The coefficient points in precisely the direction predicted by the descriptive analysis.

**E07**

escape_L2 ~ Focus + Agent + Goal_Type

```python
E07= smf.logit(
    "escape_L2 ~  Focus + Agent + Goal_Type",
    data=goals
    ).fit()

print(E07.summary())
```

Goal Type's p-value improves a little bit and the coefficient increases. The whole model is better (LLR p-value: 0.001709 vs. 0.002758) although not better than the focus-agency additive model's p (0.001351). Everything points in the same direction as in the opportunity analysis notebook.

Yet, this model shows a better picture of how counterfactual processing is shaped by considering the 3 cognitive dimensions (Focus, Agency and Goal_Type). In this sense, Agency shows a strong contribution, Focus shows a consistent positive contribution approaching significance and Goal Type shows a modest positive contribution. Focus and Goal Type still appear to move behaviour in systematic directions. 

4. Summary

- Separating the no-goal condition reduced heterogeneity in the remaining data, resulting in clearer predictor hierarchies coherent with opportunity analysis.

- Agency predicts escaping alternative L2 responses among participants who actually have a goal to protect. Agency opens the opportunity. Focus modifies where responses tend to go.  

- Goal Type consistently exhibited positive coefficients across additive models, although its independent contribution remained modest and did not reach conventional levels of statistical significance.

- Given results of the full additive model, escape responses are best understood as the joint product of agency, perspective, and goal structure, although agency provides the strongest independent contribution. Adding Goal Type preserves the pattern of effects and yields a coherent cognitive account in which agency remains the dominant predictor while focus and goal structure contribute more modestly.
