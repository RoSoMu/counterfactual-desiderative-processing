# No-goal Contexts

Previous focus structure effects (see notebooks 01 to 06) as well as further exhaustive modeling show consistently the heterogeneity of combined data; by spliting data into goal contexts and no-goal contexts, both effects as well as models improved despite the pruning of observations. 

No-goal contexts were analysed separately to determine whether the mechanisms identified in goal-oriented counterfactuals also appeared in the absence of an explicit consequent goal. Owing to the smaller number of observations, **these analyses are descriptive and intended to provide qualitative comparison rather than inferential conclusions**.

Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### No-Goal Contexts

1. Escape L2 models
2. Transfer Resolution Models
3. Correct Resolution Models

Read Data

```python
subgroup_theoretical = pd.read_csv("../../data/processed/subgroup_theoretical.csv")
```

### 1. Escape L2 Models

Create escape_2:

Binary Outcome: responses that are not "L2_other": those who managed to escape L2_other by choosing either a *correct* response or an *L1_transfer* response.

```python
subgroup_theoretical["escape_L2"] = (
    subgroup_theoretical["Response_Full"] != "L2_other"
).astype(int)
```

Gather no-goal context data

```python
no_goal = subgroup_theoretical[subgroup_theoretical["Goal_Type"]== "no_goal"].copy()
```

#### Counts and Proportions for no-goal contexts

Escape_L2 responses count for no_goals:

```python
no_goal_escape_L2_counts = pd.crosstab(
    no_goal["escape_L2"],
    no_goal["Response_Full"],
    margins= True)

no_goal_escape_L2_counts
```

Escape_L2 responses proportions for no-goal contexts:

```python
no_goal_escape_L2_props = no_goal["escape_L2"].value_counts(normalize = True)
no_goal_escape_L2_props
```

From 66 observations in the no_goal group, 81.8% of responses managed to escape L2_other options. This is a similar proportion to the total proportion of observations that managaged to escape the L2 landscape for goal contexts (80.3%)

Counts and Proportions by Condition

```python
no_goal_counts_condition = pd.crosstab(
    [no_goal["Agent"],no_goal["Focus"]],
    no_goal["escape_L2"],
    margins = True)

no_goal_counts_condition
```

```python
no_goal_props_condition = pd.crosstab(
    [no_goal["Agent"],no_goal["Focus"]],
    no_goal["escape_L2"],
    normalize= "index")

no_goal_props_condition
```

As opposed to goal-context behaviour, Agency does not seem to play the same part in the process of escaping the L2_other exploration for non-goal contexts. If anything, Focus would need to explain more than Agency.

#### Logistic Models

Import Libraries

```python
import statsmodels.formula.api as smf
import statsmodels.api as sm
from scipy.stats import chi2
```

Sanity check:

```python
no_goal["escape_L2"].value_counts()
```

**Main Effect Models**

**E12**

escape_L2 ~ Focus

```python
E12 = smf.logit(
    "escape_L2 ~ Focus",
    data=no_goal
    ).fit()

print(E12.summary())
```

**E13**

escape_L2 ~ Agent

```python
E13 = smf.logit(
    "escape_L2 ~ Agent",
    data=no_goal
    ).fit()

print(E13.summary())
```

- Even though there are only 66 observations to render anything statistically conclusive, escaping the L2 landscape for no-goal context modeling shows to be coherent with EDA. Focus (p-value = 0.142) captures more unique information than Agency (p-value = 0.809) whereas the oppostite hierarchy characterised the goal conditions.


### 2. Transfer Resolution

Create resolved_transfer binary outcome

```python
escapees = subgroup_theoretical[subgroup_theoretical["escape_L2"] == 1].copy()
```

```python
escapees["resolved_transfer"] = (escapees["Response_Full"] == "L1_transfer").astype(int)
```

Gather no-goal context data

```python
no_goal_escapees = escapees[escapees["Goal_Type"] == "no_goal"].copy()
```

```python
no_goal_escapees.shape
```

#### Counts and Proportions for no-goal contexts

```python
no_goal_resolved_transfer_counts = pd.crosstab(
    no_goal_escapees["resolved_transfer"],
    no_goal_escapees["Response_Full"],
    margins=True)

no_goal_resolved_transfer_counts
```

```python
no_goal_resolved_transfer_props = no_goal_escapees["resolved_transfer"].value_counts(normalize=True)
no_goal_resolved_transfer_props
```

From 54 observations no-goal observations that managed to escape the L2-other options, almost 63% were resolved by means of transfer. This is a slightly higher proportion than transfer resolution's proportion for goal contexts (58.4%)

Counts and Proportions by Condition

```python
no_goal_counts_condition = pd.crosstab(
    [no_goal_escapees["Goal_Type"], no_goal_escapees["Agent"], no_goal_escapees["Focus"]],
    no_goal_escapees["resolved_transfer"],
    margins = True)

no_goal_counts_condition
```

```python
no_goal_props_condition = pd.crosstab(
    [no_goal_escapees["Agent"],no_goal_escapees["Focus"]],
    no_goal_escapees["resolved_transfer"],
    normalize= "index")

no_goal_props_condition
```

Coherent with EDA, proportions hardly change once agency is introduced when Focus = I; it drastically changes, however, when Focus = They. Therefore, since we are not running interactions, Agency should be explaining more than Focus. 

#### Logistic Models

Sanity Check:

```python
no_goal_escapees["resolved_transfer"].value_counts()
```

**Main Effect Models**

```python
RT12 = smf.logit(
    "resolved_transfer ~ Focus",
    data=no_goal_escapees
    ).fit()

print(RT12.summary())
```

```python
RT13 = smf.logit(
    "resolved_transfer ~ Agent",
    data=no_goal_escapees
    ).fit()

print(RT13.summary())
```

- As expected, once no-goal context are analysed on their own, 54 observations are truly not enough to see anything significant. 

- However, consistent with previous descriptive analyses, Agency in this case seems to help better resolution by transfer (p-value= 0.355) as opposed to Focus (p-value= 0.982) whereas the oppostite hierarchy characterised the goal conditions.

### 3. Correct Resolution

Create resolved_correct binary outcome

```python
escapees["resolved_correct"] = (escapees["Response_Full"] == "correct").astype(int)
```

Gather no-goal context data

```python
no_goal_escapees = escapees[escapees["Goal_Type"] == "no_goal"].copy()
```

#### Counts and Proportions for no-goal contexts

```python
no_goal_resolved_correct_counts = pd.crosstab(
    no_goal_escapees["resolved_correct"],
    no_goal_escapees["Response_Full"],
    margins=True)

no_goal_resolved_correct_counts
```

```python
no_goal_resolved_correct_props = no_goal_escapees["resolved_correct"].value_counts(normalize=True)
no_goal_resolved_correct_props
```

From 54 observations no-goal observations that managed to escape the L2-other options, 33.3% were resolved correctly. The proportion of no-goal observations that were resolved correctly is lower than the proportion of goal context observations that were resolved correctly (40.5%). 

Counts and Proportions by Condition

```python
no_goal_counts_condition = pd.crosstab(
    [no_goal_escapees["Goal_Type"], no_goal_escapees["Agent"], no_goal_escapees["Focus"]],
    no_goal_escapees["resolved_correct"],
    margins = True)

no_goal_counts_condition
```

```python
no_goal_props_condition = pd.crosstab(
    [no_goal_escapees["Agent"],no_goal_escapees["Focus"]],
    no_goal_escapees["resolved_correct"],
    normalize= "index")

no_goal_props_condition
```

The proportion of correct responses increased slightly when agency is introduced in the antecedent when Focus = I; when Focus = They, it increases almost twice as much. However, from opportunity analysis, one subgroup (Agent = 1, Focus = They) showed unusually efficient conversion toward correct responses. Given the limited number of observations, this pattern should be interpreted cautiously.

#### Logistic Models

Sanity Check:

```python
no_goal_escapees["resolved_correct"].value_counts()
```

**Main Effect Models**

```python
RC12 = smf.logit(
    "resolved_correct ~ Focus",
    data=no_goal_escapees
    ).fit()

print(RC12.summary())
```

```python
RC13 = smf.logit(
    "resolved_correct ~ Agent",
    data=no_goal_escapees
    ).fit()

print(RC13.summary())
```

- Again, once no-goal context are analysed on their own, 54 observations are truly not enough to see anything significant. 

- In any case, Agency in this case also seems to help resolving correctly (p-value = 0.1659) as opposed to Focus (p-value= 0.6880). This also shows that Agency as a predictor for resolving correctly is better than Agency as a predictor for resolving by means of transfer (p = 0.355). Unfortunately, there aren't enough observations to make any claims.

- Looking into agency-focus interactions to explain differences in variation would have been the right way to proceed. 

In summary:

- The no-goal items on their own do not seem to merely introduce statistical noise; they appear to follow a qualitatively different behavioural strategy to goals. This explains the variability we found when analyses were carried out on the combined dataset.

- Across all exploratory analyses, no-goal contexts consistently displayed patterns that contrasted with those observed in goal-context items. This consistency across descriptive summaries and models suggests that no-goal items may reflect a different behavioural regime rather than merely contributing random variability.
