# No-goal Contexts

The following models take into account that, throughout analyses, *no_goal* seems to behave entirely differently to the other two Goal Type conditions (*goal_frequent* and *goal_non_frequent*). 

Previous focus structure effects (see notebooks 11 to 6) as well as furhter exhaustive modelling (see notebooks 11_appendix vs. 12_appendix) show consistently the heterogeneity of combined data; by spliting data into goal contexts and no-goal contexts, both effects as well as models improved despite the pruning of observations. No-goal contexts are also analysed separately; however, and although results are coherent with what is seen by comparing combined data to goal behaviour data, the sample is clearly not enough to show anything significant from no-goal contexts.

Taking into account the effects summary (notebook 5) and response opportunity analysis (notebook 6), starting from the main effect identified in the previous analyses, predictors were added only when supported by opportunity.

Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Goal Behaviour Subgroup

1. Escape L2 models
2. Transfer Resolution Models
3. Correct Resolution Models

Read Data

```python
subgroup_theoretical = pd.read_csv("../../data/processed/subgroup_theoretical.csv")
```

### 1. Escape L2 Models

Create escape_2

Binary Outcome: responses that are not "L2_other": those who managed to escape L2_other by choosing either a *correct* response or an *L1_transfer* response.

```python
subgroup_theoretical["escape_L2"] = (
    subgroup_theoretical["Response_Full"] != "L2_other"
).astype(int)
```

Split data into goals vs. no_goals

```python
goals = subgroup_theoretical[subgroup_theoretical["Goal_Type"]!= "no_goal"].copy()
```

```python
no_goal = subgroup_theoretical[subgroup_theoretical["Goal_Type"]== "no_goal"].copy()
```

**Counts and Proportions for no-goal contexts**

Escape_L2 responses count for no_goals:

```python
no_goal_escape_L2_counts = pd.crosstab(
    no_goal["escape_L2"],
    no_goal["Response_Full"],
    margins= True)

no_goal_escape_L2_counts
```

Escape_L2 responses proportions for no-goal contexts:

```python
no_goal_escape_L2_props = no_goal["escape_L2"].value_counts(normalize = True)
no_goal_escape_L2_props
```

From 66 observations in the no_goal group, 81.8% of responses managed to escape L2_other options. 

Counts and Proportions by Condition:

```python
no_goal_counts_condition = pd.crosstab(
    [no_goal["Agent"],no_goal["Focus"]],
    no_goal["escape_L2"],
    margins = True)

no_goal_counts_condition
```

```python
no_goal_props_condition = pd.crosstab(
    [no_goal["Agent"],no_goal["Focus"]],
    no_goal["escape_L2"],
    normalize= "index")

no_goal_props_condition
```

Agency doesn't seem to play the same part in the process of escaping the L2_other for non-goal contexts.

Logistic Models

Import Libraries

```python
import statsmodels.formula.api as smf
import statsmodels.api as sm
from scipy.stats import chi2
```

Sanity check:

```python
no_goal["escape_L2"].value_counts()
```

**Main Effect Models**

**E12**

escape_L2 ~ Focus

```python
E12 = smf.logit(
    "escape_L2 ~ Focus",
    data=no_goal
    ).fit()

print(E12.summary())
```

**E13**

escape_L2 ~ Agent

```python
E13 = smf.logit(
    "escape_L2 ~ Agent",
    data=no_goal
    ).fit()

print(E13.summary())
```

- Even though there are only 66 observations to render anything statistically conclusive, escaping the L2 landscape for no-goal context modelling shows to be coherent with EDA; Focus (p-value = 0.142) captured more unique information than Agency (p-value = 0.809) whereas the oppostite hierarchy characterised the goal conditions.

- Although they are underpowered becuase of sample size, the pattern is consistent with descriptive analyses.

### 2. Transfer Resolution

Create resolved_transfer binary outcome

```python
escapees = subgroup_theoretical[subgroup_theoretical["escape_L2"] == 1].copy()
```

Split escapees data into goals vs. no_goals

```python
escapees["resolved_transfer"] = (escapees["Response_Full"] == "L1_transfer").astype(int)
```

```python
goals_escapees = escapees[escapees["Goal_Type"] != "no_goal"].copy()
```

```python
no_goal_escapees = escapees[escapees["Goal_Type"] == "no_goal"].copy()
```

**Counts and Proportions for no-goal contexts**

```python
no_goal_resolved_transfer_counts = pd.crosstab(
    no_goal_escapees["resolved_transfer"],
    no_goal_escapees["Response_Full"],
    margins=True)

no_goal_resolved_transfer_counts
```

```python
no_goal_resolved_transfer_props = no_goal_escapees["resolved_transfer"].value_counts(normalize=True)
no_goal_resolved_transfer_props
```

From 54 observations no-goal observations that managed to escape the L2-other options, almost 63% were resolved by means of transfer. 

**Counts and Proportions by Condition**

```python
no_goal_counts_condition = pd.crosstab(
    [no_goal_escapees["Goal_Type"], no_goal_escapees["Agent"], no_goal_escapees["Focus"]],
    no_goal_escapees["resolved_transfer"],
    margins = True)

no_goal_counts_condition
```

```python
no_goal_props_condition = pd.crosstab(
    [no_goal_escapees["Agent"],no_goal_escapees["Focus"]],
    no_goal_escapees["resolved_transfer"],
    normalize= "index")

no_goal_props_condition
```

Logistic Models

Sanity Check:

```python
no_goal_escapees["resolved_transfer"].value_counts()
```

**Main Effect Models**

```python
RT12 = smf.logit(
    "resolved_transfer ~ Focus",
    data=no_goal_escapees
    ).fit()

print(RT12.summary())
```

```python
RT13 = smf.logit(
    "resolved_transfer ~ Agent",
    data=no_goal_escapees
    ).fit()

print(RT13.summary())
```

- As expected, once no-goal context are analysed on their own, 54 observations are truly not enough to see anything significant. However, consistent with previous descriptive analyses, Agency in this case seems to help better resolution by transfer (p-value= 0.355) as opposed to Focus (p-value= 0.982). This seems to mirror goal-context behaviour.

### 3. Correct Resolution

```python
escapees["resolved_correct"] = (escapees["Response_Full"] == "correct").astype(int)
```

Split data into goals and no_goals

```python
goals_escapees = escapees[escapees["Goal_Type"] != "no_goal"].copy()
```

```python
no_goal_escapees = escapees[escapees["Goal_Type"] == "no_goal"].copy()
```

**Counts and Proportions for no-goal contexts**

```python
no_goal_resolved_correct_counts = pd.crosstab(
    no_goal_escapees["resolved_correct"],
    no_goal_escapees["Response_Full"],
    margins=True)

no_goal_resolved_correct_counts
```

```python
no_goal_resolved_correct_props = no_goal_escapees["resolved_correct"].value_counts(normalize=True)
no_goal_resolved_correct_props
```

From 54 observations no-goal observations that managed to escape the L2-other options,  33.3% were resolved correctly. 

**Counts and Proportions by Condition**

```python
no_goal_counts_condition = pd.crosstab(
    [no_goal_escapees["Goal_Type"], no_goal_escapees["Agent"], no_goal_escapees["Focus"]],
    no_goal_escapees["resolved_correct"],
    margins = True)

no_goal_counts_condition
```

```python
no_goal_props_condition = pd.crosstab(
    [no_goal_escapees["Agent"],no_goal_escapees["Focus"]],
    no_goal_escapees["resolved_correct"],
    normalize= "index")

no_goal_props_condition
```

Logistic Models

Sanity Check:

```python
no_goal_escapees["resolved_correct"].value_counts()
```

**Main Effect Models**

```python
RC12 = smf.logit(
    "resolved_correct ~ Focus",
    data=no_goal_escapees
    ).fit()

print(RC12.summary())
```

```python
RC13 = smf.logit(
    "resolved_correct ~ Agent",
    data=no_goal_escapees
    ).fit()

print(RC13.summary())
```

- The no-goal items on their own do not seem to merely introduce statistical noise; they appear to follow a qualitatively different behavioural strategy to goals. This explains the variability we found when analyses were carried out on the combined dataset

- Across all exploratory analyses, no-goal contexts consistently displayed patterns that contrasted with those observed in goal-context items. This consistency across descriptive summaries and models suggests that no-goal items may reflect a different behavioural regime rather than merely contributing random variability.

- In any case, and as expected, Agency in this case also seems to help resolving correctly (p-value = 0.1659) as opposed to Focus (p-value= 0.6880). This also shows that Agency as a predictor for resolving correctly is better than Agency as a predictor for resolving by means of transfer (p = 0.355). Unfortunately, there aren't enough observations to make any claims.
