Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Get data from tables

```python
escape_summary = pd.read_csv("../../reports/tables/escape_L2_goal_split_summary.csv")
escape_summary.head()
```

```python
transfer_summary = pd.read_csv("../../reports/tables/transfer_goal_split_summary.csv")
transfer_summary.head()
```

```python
correct_summary = pd.read_csv("../../reports/tables/correct_goal_split_summary.csv")
correct_summary.head()
```

```python
modelling_summary = pd.concat([escape_summary, transfer_summary, correct_summary])
modelling_summary.head()
```

```python
modelling_summary.shape
```

```python
modelling_summary.columns
```

1. Analytical question:

   **What about no-goal contexts?**

2. Escaping the L2


--> Response Opportunity

```python
no_goal_summary_metrics = pd.read_csv("../../reports/tables/no_goal_summary_metrics.csv",
                                   index_col=[0, 1])
no_goal_summary_metrics
```

```python
no_goal_summary_metrics.index.names
```

```python
no_goal_summary_metrics.dtypes
```

```python
L2_suppression = no_goal_summary_metrics[["L2_suppression"]]
L2_suppression
```

```python
sns.heatmap(
    L2_suppression,
    annot=True,
    cmap="Reds",
    center=0,
    fmt=".2f",
    vmin=0,
    vmax = 0.50
)

plt.title("L2 Suppression across No-goal Contexts")
plt.ylabel("Agent – Focus")
plt.xlabel("")
plt.show()
```

As opposed to goal context behaviour, self-perspective is associated with higher suppression of L2-other options.


--> Modelling Comparison

The opportunity analysis identifies a particularly stable descriptive pattern: self-perspective  is associated with greater suppression of competing L2 responses across all Agency conditions. The sample is too small to make any claims; however, it is enough to show a mirror pattern

The modelling summary compares main predictors (Focus and Agency) in no-goal contexts to goal contexts':

- E12: escape_L2 ~ Focus (Goal_Behaviour = No_goal) vs. E01 (Goal_Behaviour = Goal)

- E13: escape_L2 ~ Agent (Goal_Behaviour = No_goal) vs. E02 (Goal_Behaviour = Goal)

**Escape responses**

```python
escape_models = modelling_summary[modelling_summary["Model"].isin(["E01", "E02", "E12", "E13"])]
```

```python
escape_data =escape_models[["Goal_Behaviour","Model", "Formula","Model_LLR_p"]]
escape_data
```

```python
plt.figure(figsize=(7, 5))

sns.lineplot(
    data=escape_data,
    x="Formula",
    y="Model_LLR_p",
    hue="Goal_Behaviour",
    marker="o"
)

plt.xticks([0, 1], ["escape_L2 ~ Focus", "escape_L2 ~ Agent"])
plt.ylim(0, 1)

plt.xlabel("Model Formula")
plt.ylabel("Model LLR p-value")
plt.title("Escape L2: Model LLR p-value no-goal vs goal comparison")

plt.legend(title="Goal Behaviour")
plt.show()
```

--> Modelling Comparison

The opportunity analysis identifies a particularly stable descriptive pattern: Agency

The modelling summary compares main predictors (Focus and Agency) in no-goal contexts to goal contexts':

- RT12: resolved_tranfer ~ Focus (Goal_Behaviour = No_goal) vs. RT01 (Goal_Behaviour = Goal)

- RT13: resolved_transfer ~ Agent (Goal_Behaviour = No_goal) vs. RT02 (Goal_Behaviour = Goal)

- RC12: resolved_tranfer ~ Focus (Goal_Behaviour = No_goal) vs. RC01 (Goal_Behaviour = Goal)

- RC13: resolved_transfer ~ Agent (Goal_Behaviour = No_goal) vs. RC02 (Goal_Behaviour = Goal)

**Transfer responses**

```python
transfer_models = modelling_summary[modelling_summary["Model"].isin(["RT01", "RT02", "RT12", "RT13"])]
```

```python
transfer_data =transfer_models[["Goal_Behaviour","Model", "Formula","Model_LLR_p"]]
transfer_data
```
