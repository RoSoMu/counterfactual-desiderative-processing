**Escape responses**

```python
escape_models = modelling_summary[modelling_summary["Model"].isin(["E01", "E02", "E12", "E13"])]
```

```python
escape_data =escape_models[["Goal_Behaviour","Model", "Formula","Model_LLR_p"]]
escape_data
```

```python
escape_data["Goal_Behaviour"] = escape_data["Goal_Behaviour"].replace({
    "Goal": "Goal",
    "No_goal": "No goal"
})
```

```python
plt.figure(figsize=(7, 5))

ax = sns.lineplot(
    data=escape_data,
    x="Formula",
    y="Model_LLR_p",
    hue="Goal_Behaviour",
    marker="o"
)

ax.set_xticks([0, 1])
ax.set_xticklabels(["Focus", "Agency"])
ax.set_ylim(0, 1)
ax.set_xlabel("Predictor")
ax.set_ylabel("Likelihood-ratio test p-value")
ax.set_title(
    "Escape L2: Goal vs. No-goal Contexts"
)

ax.legend(
    title="Context"
)

# Significance reference

ax.axhline(
    0.05,
    linestyle="--",
    linewidth=1,
    alpha=0.6
)

# Add p-value labels

for _, row in escape_data.iterrows():
    x = 0 if "Focus" in row["Formula"] else 1
    y = row["Model_LLR_p"]
    if y < 0.05:
        offset = 0.025
    elif row["Goal_Behaviour"] == "No goal":
        offset = 0.025
    else:
        offset = -0.035
    ax.text(
        x,
        y + offset,
        f"{y:.3f}",
        ha="center"
    )
```

The behavioural message is simple:

* For goal contexts: Focus is weak  (p ≈ .139) and Agency is strong (p ≈ .0016).
* For no-goal contexts: Focus is about the same (p ≈ .142) but Agency collapses (p ≈ .808).

In goal contexts, evidence for Escape_L2 was concentrated in Agency rather than Focus. In no-goal contexts, this asymmetry disappeared: evidence for Focus remained virtually unchanged, whereas evidence for Agency collapsed.

This also provides further explaining on the type of variation that was added in the combined dataset: the no-goal observations weren’t merely adding noise. They may have been adding observations generated under a different contextual organisation of the predictors.

**Transfer responses**

```python
transfer_models = modelling_summary[modelling_summary["Model"].isin(["RT01", "RT02", "RT08", "RT12", "RT13", "RT15"])]
```

```python
transfer_data = transfer_models[["Goal_Behaviour","Model", "Formula","Model_LLR_p"]].drop_duplicates()
transfer_data
```

```python
transfer_data["Goal_Behaviour"] = transfer_data["Goal_Behaviour"].replace({
    "Goal": "Goal",
    "No_goal": "No goal"
})
```

```python
plt.figure(figsize=(7, 5))

ax = sns.lineplot(
    data=transfer_data,
    x="Formula",
    y="Model_LLR_p",
    hue="Goal_Behaviour",
    marker="o"
)

ax.set_xticks([0, 1, 2])
ax.set_xticklabels(["Focus", "Agency", "Focus x Agency"])
ax.set_ylim(0, 1)
ax.set_xlabel("Predictor")
ax.set_ylabel("Likelihood-ratio test p-value")
ax.set_title(
    "Transfer resolution: Goal vs. No-goal Contexts"
)

ax.legend(
    title="Context"
)

# Significance reference

ax.axhline(
    0.05,
    linestyle="--",
    linewidth=1,
    alpha=0.6
)

# Add p-value labels


for _, row in transfer_data.iterrows():
    if row["Formula"] == "resolved_transfer ~ Focus":
        x = 0
    elif row["Formula"] == "resolved_transfer ~ Agent":
        x = 1
    else:  # Focus * Agent
        x = 2
    y = row["Model_LLR_p"]
    if y < 0.05:
        offset = 0.025
    elif row["Goal_Behaviour"] == "No_goal":
        offset = 0.025
    else:
        offset = -0.035
    ax.text(
        x,
        y + offset,
        f"{y:.3f}",
        ha="center"
    )
```

Reomving the goal doesn't mereley make evryting weaker because of losing power, it is the relative structure that changes. 

In goal contexts, Focus and Focus × Agency are the two models showing the strongest evidence of structure, while Agency alone contributes very little. In no-goal contexts, Focus essentially disappears as an explanatory dimension altogether: .060 → .983. And then Agency and the interaction end up virtually together: .355 and .341.

In the goal data, So once Focus is allowed to participate in the Agency relationship, the model evidence changes dramatically. That’s completely consistent with what we saw in the RT11 predicted-probability plot: Agency did not have one uniform relationship with transfer resolution; what it did depended on Focus.

but in no-goal, almost nothing happens when Focus is broguht into Agency strucutre.

For both resolution outcomes, when a goal is present, Focus carries much more evidence than Agency alone. Remove the goal, and that Focus signal collapses:

* Transfer: .060 → .983
* Correct: .087 → .688

Meanwhile Agency goes in the other direction:

* Transfer: .502 → .355
* Correct: .615 → .166

Goal removal is associated with a reversal in the relative explanatory profile of Focus and Agency. In goal contexts, resolution is more strongly structured by Focus than by Agency; in no-goal contexts, the relative pattern reverses, with Focus providing little evidence while Agency becomes comparatively more informative. This pattern is reproduced across both transfer and correct resolution.

The interactions tells us something extra about the goal condition that the other modesl cannot:
for goals, adding the Focus–Agency structure pulls the model back toward the Focus side
for no- goals, he interaction no longer returns toward the Focus-model evidence in the same way.

4. No-goal vs Goal contexts synthesis

Separating goal from no-goal contexts suggests that part of the variability observed in the combined dataset resulted from averaging across qualitatively different behavioural structures. In goal contexts, Agency strongly organised L2 escape, whereas Focus was comparatively stable across goal and no-goal conditions. At the resolution stage, however, goal contexts showed a comparatively stronger Focus-based structure than Agency alone for both transfer and correct resolution. Once the goal was removed, this organisation changed: the Focus signal was greatly attenuated, while Agency became comparatively more informative. Thus, no-goal trials did not simply weaken the patterns observed in goal contexts; they redistributed the relative contribution of Focus and Agency across stages of behaviour. Models fitted to the combined data were therefore required to accommodate partially opposing or differently organised trends, converting structured context-dependent variation into apparently weak or inconsistent average effects.

The behavioural sequence look like:
In goal contexts: Agency strongly structures whethere L2 is escaped and Focus becomes comparatively prominent in how the escape is resolved.
In no-goal context: Agency loses that strong escape relationship and Focus also loses teh resolutiton structure seen under goals.
That makes Goal Behaviour look less like another predictor affecting response probability and more like a context that changes the organisation of the processing landscape.

The no-goal analysis helps explain the substantial variability encountered in models fitted to the combined dataset. Rather than contributing random dispersion around the patterns observed in goal contexts, no-goal trials showed a different organisation of L2 escape and subsequent resolution. Agency strongly differentiated escape behaviour when a goal was available but not when it was absent, whereas the resolution models showed comparatively stronger Focus-related structure under goal contexts for both transfer and correct outcomes; this Focus pattern was markedly attenuated following goal removal. Pooling these contexts therefore required a single model to average across effects whose relative strength—and in some cases behavioural direction—changed with goal availability. What appeared as weak or inconsistent effects in the combined analysis may consequently reflect structured context-dependent heterogeneity rather than the absence of systematic behaviour.
