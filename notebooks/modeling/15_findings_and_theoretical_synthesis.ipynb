# Findings and Theoretical Synthesis


This notebook returns to the predictions established in `00_theoretical_framework_and_predictions.ipynb` and evaluates them against the behavioural patterns identified across the preceding analyses.

The original framework proposed that properties of the third-conditional context could facilitate learners' subsequent selection of the Past Perfect in a desiderative sentence.

Three directional predictions were made:

- **Goal Type:** frequent goal > non-frequent goal > no goal
- **Agency:** explicit agency > implicit agency
- **Focus:** self-focus (`I`) > other-focus (`They`)

Together, these predictions identified a theoretically most favourable configuration:

> **Frequent goal + Explicit agency + Self-focus (`I`)**

and a theoretically least favourable configuration:

> **No goal + Implicit agency + Other-focus (`They`)**

These were predicted endpoints rather than a complete ranking of all experimental combinations.

The analyses that followed, however, allowed the response process to be examined in greater detail: not only whether participants selected the target form, but how the experimental manipulations related to the opportunity to move beyond competing L2 responses and, once that opportunity emerged, to the subsequent resolution of the response.

The observed behaviour did not resolve into this single ordered continuum. Instead, Goal, Agency, and Focus were associated with different aspects of the response process, and their relationships changed across experimental contexts.

## From predicted facilitation to observed response proceses

The original hypotheses predicted which counterfactual contexts would provide more favourable conditions for subsequent target Past Perfect selection.

The modelling analyses subsequently allowed this behavioural outcome to be examined as a two-stage process.

### Stage 1 — L2 escape: reaching a resolution opportunity

The first stage asked whether participants moved beyond the competing `L2_other` responses.

At this stage, the question was not yet whether the response ultimately resolved toward L1 transfer or toward the target form. It was whether the counterfactual context was associated with a greater opportunity for that
resolution to occur.

### Stage 2 — Resolution: where did the response go?

Among responses that moved beyond the competing L2 alternatives, the second stage examined the destination of resolution:

- **L1-transfer resolution**
- **Correct/target resolution**

This distinction made it possible to ask a question that the original prediction of overall facilitation could not answer:

**Did a theoretically favourable counterfactual context make resolution more likely, make correct resolution more likely, or both?**

## Stage 1 — L2 escape: reaching a resolution opportunity

The **descriptive opportunity analysis** already showed the clearest repeated pattern: moving from implicit to explicit Agency increased L2 suppression/escape in every goal-context combination:

* frequent, I: .15 → .38
* frequent, They: .17 → .50
* non-frequent, I: .17 → .44
* non-frequent, They: .44 → .50

The model (E07) asks whether that structure survives when **Focus + Agency + Goal Type are considered together** and it did: 
* Agency is the clearest contribution: coefficient 1.8656, p=.004 
* Focus and Goal Type shift the probability landscape in consistent directions but with weaker individual evidence.

#### Predicted probability of L2 escape

The E07 model estimated the probability of escaping the competing `L2_other` response space for each combination of Focus, Agency, and Goal Type.

| Goal Type | Focus | Implicit agency | Explicit agency |
|---|---|---:|---:|
| Frequent | Self-focus (`I`) | 58.4% | **90.1%** |
| Frequent | Other-focus (`They`) | 77.4% | **95.7%** |
| Non-frequent | Self-focus (`I`) | 72.7% | **94.5%** |
| Non-frequent | Other-focus (`They`) | 86.7% | **97.7%** |

Across all four experimental configurations, explicit agency was associated with a higher predicted probability of L2 escape than implicit agency. The largest difference occurred in the frequent-goal, self-focused context (58.4% → 90.1%).

Focus and Goal Type further shifted the probability landscape: other-focused contexts had higher predicted probabilities than their self-focused counterparts, and non-frequent goals had higher predicted probabilities than their frequent-goal counterparts. Neither pattern overturned the consistent Agency distinction.

**At Stage 1, therefore, explicit agency was the clearest factor organising the move away from competing L2 responses and toward a resolution opportunity. This does not yet tell us where that resolution went.**

## Stage 2 — Resolution: where did the response go?

### Transfer resolution

The **descriptive opportunity analysis** showed that the proportion of resolution opportunities shifting toward L1 transfer depended strongly on the experimental configuration:

- frequent goal, `I`: **.01 → .19** from implicit to explicit agency
- frequent goal, `They`: **.14 → .38** from implicit to explicit agency
- non-frequent goal, `I`: **.04 → .44** from implicit to explicit agency
- non-frequent goal, `They`: **.53 → .38** from implicit to explicit agency

For transfer, the descriptive opportunity analysis measured L1_shift. These descriptive metrics capture changes associated with the available resolution opportunity and should not be read as the model-predicted probabilities presented below.

Unlike Stage 1, moving from implicit to explicit agency did **not** produce the same directional change in every configuration. The largest contrast appeared in the non-frequent, other-focused condition, where transfer was already high under implicit agency and decreased under explicit agency.

The model (RT11) asks whether this conditional structure survives when **Focus × Agency + Goal Type are considered together**:

- Under **implicit agency**, other-focus (`They`) is associated with greater predicted transfer resolution than self-focus (`I`).
- Explicit agency changes that Focus distinction: predicted transfer increases under self-focus while decreasin slightly under other-focus, substantially reducing the difference between the two perspectives.
- **Non-frequent goals** shift predicted transfer resolution upward while broadly preserving the underlying Focus × Agency structure.

#### Predicted probability of transfer resolution

The RT11 model estimated the probability of **L1-transfer resolution**, given a resolution opportunity, for each combination of Focus, Agency, and Goal Type.

| Goal Type | Focus | Implicit agency | Explicit agency |
|---|---|---:|---:|
| Frequent | Self-focus (`I`) | 33% | **55%** |
| Frequent | Other-focus (`They`) | **65%** | 54% |
| Non-frequent | Self-focus (`I`) | 48% | **70%** |
| Non-frequent | Other-focus (`They`) | **78%** | 69% |

Under **implicit agency**, other-focused contexts showed substantially higher predicted probabilities of transfer resolution than their self-focused counterparts: 65% vs. 33% for frequent goals and 78% vs. 48% for non-frequent
goals.

Explicit agency changed this relationship. Under self-focus, predicted transfer increased from 33% to 55% for frequent goals and from 48% to 70% for non-frequent goals. Under other-focus, however, it decreased slightly: from 65% to 54% for frequent goals and from 78% to 69% for non-frequent goals.

**At the transfer-resolution stage, therefore, Agency did not operate as a uniform facilitator. Its relationship with transfer depended on Focus: explicit agency substantially reduced the Focus distinction that was visible under implicit agency.**

### Correct resolution

The **descriptive opportunity analysis** showed the complementary destination of resolution:

- frequent goal, `I`: **.14 → .19** from implicit to explicit agency
- frequent goal, `They`: **.03 → .12** from implicit to explicit agency
- non-frequent goal, `I`: **.12 → .00** from implicit to explicit agency
- non-frequent goal, `They`: **−.08 → .12** from implicit to explicit agency

Here again, Agency did not produce a uniform directional shift across all experimental configurations. Instead, the relationship between Agency and correct resolution depended on Focus and Goal Type.

For correct resolution, the descriptive opportunity analysis measured measured Correct_gain. These descriptive metrics capture changes associated with the available resolution opportunity and should not be read as the model-predicted probabilities presented below.

The model (RC11) asks whether this conditional structure survives when **Focus × Agency + Goal Type are considered together**:

- Under **implicit agency**, self-focus (`I`) is associated with greater predicted correct resolution than other-focus (`They`).
- Explicit agency changes that Focus distinction: predicted correct resolution decreases under self-focus while increasing under other-focus, substantially reducing the difference between the two perspectives.
- **Non-frequent goals** shift predicted correct resolution downward while broadly preserving the underlying Focus × Agency structure.

## Which context was more favourable?

The two modelling stages complicate the original idea of a single “most favourable” counterfactual environment. In the frequent-goal, self-focused condition, explicit and implicit agency were associated with different advantages at different stages of the response process.

With explicit agency (If I had saved money…), the predicted probability of escaping the competing L2 response space was substantially higher (90%) than with implicit agency (If savings had been sufficient…; 58%). Explicit agency therefore showed a clear escape advantage towards resolution.

Once a resolution opportunity had been reached, however, the relationship changed. Under implicit agency, correct resolution was predicted at 63%, compared with 33% for transfer. Under explicit agency, resolution was considerably more balanced: 45% correct and 55% transfer. Implicit agency therefore showed a conditional correct-resolution advantage, despite being associated with a lower probability of reaching that stage.

**Neither configuration can therefore be identified as globally optimal from these models.** The context most strongly associated with reaching a resolution opportunity was not the context most strongly associated with correct resolution once that opportunity existed. The original notion of a single favourable environment is therefore better understood as **a response pathway in which different properties of the counterfactual may be advantageous at different stages.**

Because the escape and resolution probabilities were estimated by separate models representing different stages of behaviour, **they should not be multiplied to construct an informal overall probability or used to rank the configurations globally.**

## Theoretical framework revisited

### Returning to processing cost: Gibson

The original framework drew on Gibson's resource-based account of linguistic processing to predict that more accessible counterfactual representations would leave greater processing resources available for the difficult L2 tense mapping.

The findings do not require this processing account to be abandoned. At the first stage of behaviour, explicit agency was strongly associated with reaching a resolution opportunity in goal-directed contexts. Sentences such as:

> *If I had saved money, I would have gone on holidays.*

provide an overt animate agent and a relatively transparent agent → action → outcome structure. Their implicit-agency counterparts:

> *If savings had been sufficient, I would have gone on holidays.*

require the antecedent to be organised around a state or circumstance rather than around the participant as its overt agent.

Importantly, however, implicit agency did not remove the participant from the counterfactual representation. Focus remained overtly encoded in the consequent (*I would have...* / *they would have...*). What changed was where the participant entered the event structure and whether that participant was represented as the explicit causal agent of the antecedent.

The resolution models therefore qualify a simple equation between processing accessibility and successful target resolution. In the frequent, self-focused condition, explicit agency was associated with a much higher predicted probability of L2 escape, yet it did not produce the highest predicted probability of correct resolution once a resolution opportunity had been reached. The implicit-agency configuration showed the opposite trade-off: lower predicted escape, but a stronger conditional tendency toward correct resolution.

**Computational economy, therefore, is not itself a resolution destination.**
A linguistic configuration may facilitate one stage of the response process without determining how the response is subsequently resolved.

This distinction also changes what it means to ask whether a counterfactual representation is "cheap". Processing cost can be considered with respect to the linguistic representation itself, but the behavioural consequences of that representation depend on the stage of processing being examined and on the participant's position within the represented event. The relevant question is therefore no longer simply whether explicit agency is computationally less demanding, but **less demanding for what subsequent operation, and within whose counterfactual representation?**

### Returning to functional counterfactuality: what was the counterfactual for?

The original prediction derived from the functional account of counterfactual thinking was straightforward: counterfactuals organised around familiar, repeatable goals were expected to provide the most favourable context for subsequent target selection.

The findings support the importance of goal structure, but not as a single continuum from more to less favourable performance.

Within goal-directed contexts, Goal Type consistently shifted the predicted destination of resolution. Non-frequent goals were associated with greater transfer resolution and lower correct resolution than frequent goals, while broadly preserving the underlying Focus × Agency organisation.

Compare, for example:

> *If I had saved money, I would have gone on holidays.*

with:

> *If I had created an online platform, I would have saved my family's business.*

Both counterfactuals represent a blocked goal and an alternative means by which a different outcome could have been obtained. They differ, however, in the kind of goal being represented. Saving money for a holiday describes a relatively familiar and repeatable means–goal relationship. Creating an online platform to save a family business represents a less routine, more consequential and potentially more personally loaded alternative.

The present experiment did not measure emotional involvement, regret, responsibility or perceived consequences. These dimensions therefore cannot be identified as mechanisms underlying the observed Goal Type effect. Nevertheless, the findings caution against treating goal presence or goal frequency as a simple measure of facilitation. A counterfactual may be highly goal-directed and functionally meaningful while also placing different demands on subsequent L2 resolution.

##### And what happens when the goal disappears?

The no-goal contexts suggest that removing explicit goal structure did more than reduce the strength of the patterns observed in goal-directed counterfactuals.

Consider the progression from:

> *If I had saved money, I would have gone on holidays.*

to:

> *If savings had been sufficient, I would have gone on holidays.*

and finally:

> *If there had been a rainstorm, I would have felt relaxed.*

The last configuration is not simply a weaker version of the first two. Its event structure is different. Rather than representing an agent performing an action as an alternative means toward a blocked goal, the antecedent introduces an external circumstance, while the person represented in the consequent functions as the experiencer of an internal state.

This distinction was reflected in the behavioural analyses. In goal-directed contexts, Agency strongly organised the probability of reaching a resolution opportunity. In the no-goal contexts, that Agency signal was substantially attenuated. At the resolution stage, the relative contribution of Focus and Agency also changed rather than simply becoming uniformly weaker.

The no-goal trials therefore appear better understood as a qualitatively different counterfactual configuration than as the low end of a single Goal-Type scale.

From a functional perspective, this distinction is meaningful. Counterfactual thinking is not defined merely by imagining an alternative past; its usefulness can depend on the relationship between a blocked outcome, the alternative means represented, and the possibility of using that information in subsequent behaviour.

The present findings are compatible with that broader functional view. What they do not support is the stronger assumption that greater functional relevance should translate monotonically into greater target-form accuracy. Goal structure helped organise the behavioural landscape, but its consequences depended on which stage and destination of the response process was examined.

### Returning to perspective: reading actors and reading readers

The findings also invite a return to the actor–reader distinction discussed by Girotto et al. (2007).

In their work, the alternatives generated by people who had themselves experienced an unsuccessful problem-solving episode differed from those generated by readers considering another person's experience. Actors tended to alter features or means involved in their own unsuccessful behaviour, whereas readers more readily undid the protagonist's choice.

The present experiment did not reproduce this distinction directly. Participants in both Focus conditions were readers of experimentally provided counterfactual scenarios. The self-focused condition (`I`) therefore did not give participants the autobiographical experience available to Girotto et al.'s actors, and the other-focused condition (`They`) should not be treated as a direct equivalent of their reader condition.

Nevertheless, the present findings suggest that perspective remained relevant even within this entirely linguistic manipulation.

Consider the frequent-goal contexts under implicit agency:

> *If savings had been sufficient, I would have gone on holidays.*

> *If savings had been sufficient, they would have gone on holidays.*

The antecedent is identical. The counterfactual circumstance is identical. What changes is the person whose unrealised outcome is represented in the consequent.

Yet once a resolution opportunity had been reached, these two perspectives were associated with markedly different predicted resolution patterns. Under implicit agency, the self-focused configuration showed greater predicted correct resolution and lower transfer, whereas the other-focused configuration showed the complementary tendency toward greater transfer and lower correct resolution.

When agency became explicit:

> *If I had saved money, I would have gone on holidays.*

> *If they had saved money, they would have gone on holidays.*

the Focus distinction in resolution was substantially reduced.

This pattern makes it difficult to treat Focus as a simple pronoun contrast. Its relationship with resolution depended on how the individual was positioned within the counterfactual event structure. When the antecedent did not overtly identify that individual as its causal agent, self- and other-focused representations were strongly differentiated. When the individual became the explicit antecedent agent, that difference largely converged.

The findings are therefore compatible with the broader insight motivating Girotto et al.'s actor–reader distinction: counterfactual reasoning depends not only on the objective event being represented, but also on the perspective from which information about that event is available and organised.

This should not be interpreted as a replication of the actor–reader effect. The present participants did not differ in actual experience of the event, and the experiment did not measure which alternatives they mentally generated, which event features they considered mutable, or whether they attributed responsibility differently across conditions.

The relationship is therefore interpretive rather than evidential: the present Focus × Agency pattern shows that linguistic perspective and causal role were jointly associated with resolution behaviour, while Girotto et al. provide a useful framework for asking why the information available to a counterfactual reasoner may depend on their position within the represented event.

One particularly interesting consequence is that autobiographical participation was not necessary for a Focus distinction to emerge in the present task. Participants remained readers in both conditions, yet representing the counterfactual from `I` rather than `They` was associated with different resolution behaviour, particularly under implicit agency.

In this limited sense, the experiment suggests a distinction between **self-focused readers and other-focused readers**: not actors and readers in Girotto et al.'s original experiential sense, but readers whose linguistic representation positioned them differently with respect to the unrealised event.

This interpretation also requires caution because the available observations were not equally distributed across the two Focus conditions. The self-focused condition contributed more observations than the other-focused condition. The apparent strength and stability of the self-focused patterns should therefore not be taken as evidence that self-focus is intrinsically more influential; the other-focused estimates were supported by fewer observations.
