This notebook returns to the predictions established in `00_theoretical_framework_and_predictions.ipynb` and evaluates them against the behavioural patterns identified across the preceding analyses.

The original framework proposed that properties of the third-conditional context could facilitate learners' subsequent selection of the Past Perfect in a desiderative sentence.

Three directional predictions were made:

- **Goal Type:** frequent goal > non-frequent goal > no goal
- **Agency:** explicit agency > implicit agency
- **Focus:** self-focus (`I`) > other-focus (`They`)

Together, these predictions identified a theoretically most favourable configuration:

> **Frequent goal + Explicit agency + Self-focus (`I`)**

and a theoretically least favourable configuration:

> **No goal + Implicit agency + Other-focus (`They`)**

These were predicted endpoints rather than a complete ranking of all experimental combinations.

The analyses that followed, however, allowed the response process to be examined in greater detail: not only whether participants selected the target form, but how the experimental manipulations related to the opportunity to move beyond competing L2 responses and, once that opportunity emerged, to the subsequent resolution of the response.

The observed behaviour did not resolve into this single ordered continuum. Instead, Goal, Agency, and Focus were associated with different aspects of the response process, and their relationships changed across experimental contexts.

### **A note on the models used in this synthesis**

The preceding modelling notebooks distinguished between broader models representing the experimental design and more parsimonious accounts supported by model comparison.

This synthesis deliberately returns to the broader models when translating the findings back into the experimental framework. This choice does not imply that every term included in those models received independent statistical
support, nor does it override the model-comparison results reported in the preceding analyses.

The purpose here is different. The experiment was theoretically constructed around **Goal Type, Focus, and Agency**, and the broader models preserve those dimensions simultaneously when estimating the behavioural landscape across the experimental configurations.

Their predicted probabilities are therefore used here as a descriptive bridge between the statistical analyses and the counterfactual contexts that participants actually encountered. Inferential conclusions remain constrained by the evidence reported in the individual modelling notebooks.

In other words, the parsimonious models identify which statistical structure was required by the data; the broader models allow the resulting behaviour to be examined within the theoretical structure that motivated the experiment.

## 1. From predicted facilitation to observed response proceses

The original hypotheses predicted which counterfactual contexts would provide more favourable conditions for subsequent target Past Perfect selection.

The modelling analyses subsequently allowed this behavioural outcome to be examined as a two-stage process.

### Stage 1 — L2 escape: reaching a resolution opportunity

The first stage asked whether participants moved beyond the competing `L2_other` responses.

At this stage, the question was not yet whether the response ultimately resolved toward L1 transfer or toward the target form. It was whether the counterfactual context was associated with a greater opportunity for that
resolution to occur.

### Stage 2 — Resolution: where did the response go?

Among responses that moved beyond the competing L2 alternatives, the second stage examined the destination of resolution:

- **L1-transfer resolution**
- **Correct/target resolution**

This distinction made it possible to ask a question that the original prediction of overall facilitation could not answer:

**Did a theoretically favourable counterfactual context make resolution more likely, make correct resolution more likely, or both?**

- non-frequent goal, `I`: **.04 → .44** from implicit to explicit agency
- non-frequent goal, `They`: **.53 → .38** from implicit to explicit agency

Unlike Stage 1, moving from implicit to explicit agency did **not** produce the same directional change in every configuration. The largest contrast appeared in the non-frequent, other-focused condition, where transfer was already high under implicit agency and decreased under explicit agency.

## 4. Which context was more favourable?

The two modelling stages complicate the original idea of a single “most favourable” counterfactual environment. In self-focused goal contexts, explicit and implicit agency were associated with different patterns at different stages of the response process.

In the frequent-goal condition, explicit agency:

> *If I had saved money, I would have gone on holidays.*

was associated with a substantially higher predicted probability of moving beyond the competing `L2_other` response space and reaching a resolution opportunity (90.1%) than implicit agency:

> *If savings had been sufficient, I would have gone on holidays.*

for which the corresponding probability was 58.4%.

Once a resolution opportunity had been reached, however, the relationship changed. Under implicit agency, correct resolution was predicted at 63%, compared with 33% for transfer. Under explicit agency, resolution was
considerably more balanced: 45% correct and 55% transfer.

The same cross-stage ordering is visible in the non-frequent, self-focused condition. Implicit agency was again associated with a lower predicted probability of reaching resolution (72.7% vs. 94.5%), but with a higher predicted probability of correct resolution once that opportunity had been reached (49% vs. 32%).

| Goal Type | Agency | Representative counterfactual | L2 escape / resolution opportunity | Transfer resolution | Correct resolution |
|---|---|---|---:|---:|---:|
| Frequent | **Implicit** | *If savings had been sufficient, I would have gone on holidays.* | **58.4%** | 33% | **63%** |
| Frequent | Explicit | *If I had saved money, I would have gone on holidays.* | 90.1% | 55% | 45% |
| Non-frequent | **Implicit** | *If the online platform had been created, I would have saved my family's business.* | **72.7%** | 48% | **49%** |
| Non-frequent | Explicit | *If I had created an online platform, I would have saved my family's business.* | 94.5% | 70% | 32% |

Within the self-focused goal conditions, therefore, the configurations with the lower predicted probabilities of leaving the `L2_other` response space were also those with the higher predicted probabilities of correct resolution once resolution occurred. Conversely, the configurations associated with more frequent movement into resolution were not those associated with the strongest conditional tendency toward the target response.

This cross-stage alignment does not establish that remaining within the `L2_other` response space represents greater exploration, additional available processing resources, or more extensive consideration of L2 alternatives. The present offline response data cannot distinguish among these possible processes. It does, however, raise the possibility that movement beyond competing L2 responses and successful target resolution are not simply two successive measures of the same underlying facilitation process.

Neither configuration can therefore be identified as globally optimal from these models. Rather, the findings suggest two potentially different properties of a favourable environment: facilitating movement toward a resolution opportunity and, conditional on reaching that opportunity, facilitating correct resolution.

Because the escape and resolution probabilities were estimated by separate models representing different stages of behaviour, they should not be multiplied to construct an informal overall probability or used to rank the configurations globally.

## 5. Theoretical framework revisited

### Returning to processing cost

The original framework drew on Gibson's resource-based account of linguistic processing to predict that more accessible counterfactual representations would leave greater processing resources available for the difficult L2 tense mapping.

The findings do not require this processing account to be abandoned. At the first stage of behaviour, explicit agency was strongly associated with reaching a resolution opportunity in goal-directed contexts. Sentences such as:

> *If I had saved money, I would have gone on holidays.*

provide an overt animate agent and a relatively transparent agent → action → outcome structure. Their implicit-agency counterparts:

> *If savings had been sufficient, I would have gone on holidays.*

require the antecedent to be organised around a state or circumstance rather than around the participant as its overt agent.

Importantly, however, implicit agency did not remove the participant from the counterfactual representation. Focus remained overtly encoded in the consequent (*I would have...* / *they would have...*). What changed was where the participant entered the event structure and whether that participant was represented as the explicit causal agent of the antecedent.

The resolution models therefore qualify a simple equation between processing accessibility and successful target resolution. In the frequent, self-focused condition, explicit agency was associated with a much higher predicted probability of L2 escape, yet it did not produce the highest predicted probability of correct resolution once a resolution opportunity had been reached. The implicit-agency configuration showed the opposite trade-off: lower predicted escape, but a stronger conditional tendency toward correct resolution.

**Computational economy, therefore, is not itself a resolution destination.**
A linguistic configuration may facilitate one stage of the response process without determining how the response is subsequently resolved.

This distinction also changes what it means to ask whether a counterfactual representation is "cheap". Processing cost can be considered with respect to the linguistic representation itself, but the behavioural consequences of that representation depend on the stage of processing being examined and on the participant's position within the represented event. The relevant question is therefore no longer simply whether explicit agency is computationally less demanding, but **less demanding for what subsequent operation, and within whose counterfactual representation?**

### Returning to functional counterfactuality: what was the counterfactual for?

The original prediction derived from the functional account of counterfactual thinking was straightforward: counterfactuals organised around familiar, repeatable goals were expected to provide the most favourable context for subsequent target selection.

The findings support the importance of goal structure, but not as a single continuum from more to less favourable performance.

Within goal-directed contexts, Goal Type consistently shifted the predicted destination of resolution. Non-frequent goals were associated with greater transfer resolution and lower correct resolution than frequent goals, while broadly preserving the underlying Focus × Agency organisation.

Compare, for example:

> *If I had saved money, I would have gone on holidays.*

with:

> *If I had created an online platform, I would have saved my family's business.*

Both counterfactuals represent a blocked goal and an alternative means by which a different outcome could have been obtained. They differ, however, in the kind of goal being represented. Saving money for a holiday describes a relatively familiar and repeatable means–goal relationship. Creating an online platform to save a family business represents a less routine, more consequential and potentially more personally loaded alternative.

The present experiment did not measure emotional involvement, regret, responsibility or perceived consequences. These dimensions therefore cannot be identified as mechanisms underlying the observed Goal Type effect. Nevertheless, the findings caution against treating goal presence or goal frequency as a simple measure of facilitation. A counterfactual may be highly goal-directed and functionally meaningful while also placing different demands on subsequent L2 resolution.

##### And what happens when the goal disappears?

The no-goal contexts suggest that removing explicit goal structure did more than reduce the strength of the patterns observed in goal-directed counterfactuals.

Consider the progression from:

> *If I had saved money, I would have gone on holidays.*

to:

> *If savings had been sufficient, I would have gone on holidays.*

and finally:

> *If there had been a rainstorm, I would have felt relaxed.*

The last configuration is not simply a weaker version of the first two. Its event structure is different. Rather than representing an agent performing an action as an alternative means toward a blocked goal, the antecedent introduces an external circumstance, while the person represented in the consequent functions as the experiencer of an internal state.

This distinction was reflected in the behavioural analyses. In goal-directed contexts, Agency strongly organised the probability of reaching a resolution opportunity. In the no-goal contexts, that Agency signal was substantially attenuated. At the resolution stage, the relative contribution of Focus and Agency also changed rather than simply becoming uniformly weaker.

The no-goal trials therefore appear better understood as a qualitatively different counterfactual configuration than as the low end of a single Goal-Type scale.

From a functional perspective, this distinction is meaningful. Counterfactual thinking is not defined merely by imagining an alternative past; its usefulness can depend on the relationship between a blocked outcome, the alternative means represented, and the possibility of using that information in subsequent behaviour.

The present findings are compatible with that broader functional view. What they do not support is the stronger assumption that greater functional relevance should translate monotonically into greater target-form accuracy. Goal structure helped organise the behavioural landscape, but its consequences depended on which stage and destination of the response process was examined.

### Returning to perspective: reading actors and reading readers

The findings also invite a return to the actor–reader distinction discussed by Girotto et al. (2007).

In their work, the alternatives generated by people who had themselves experienced an unsuccessful problem-solving episode differed from those generated by readers considering another person's experience. Actors tended to alter features or means involved in their own unsuccessful behaviour, whereas readers more readily undid the protagonist's choice.

The present experiment did not reproduce this distinction directly. Participants in both Focus conditions were readers of experimentally provided counterfactual scenarios. The self-focused condition (`I`) therefore did not give participants the autobiographical experience available to Girotto et al.'s actors, and the other-focused condition (`They`) should not be treated as a direct equivalent of their reader condition.

Nevertheless, the present findings suggest that perspective remained relevant even within this entirely linguistic manipulation.

Consider the frequent-goal contexts under implicit agency:

> *If savings had been sufficient, I would have gone on holidays.*

> *If savings had been sufficient, they would have gone on holidays.*

The antecedent is identical. The counterfactual circumstance is identical. What changes is the person whose unrealised outcome is represented in the consequent.

Yet once a resolution opportunity had been reached, these two perspectives were associated with markedly different predicted resolution patterns. Under implicit agency, the self-focused configuration showed greater predicted
correct resolution and lower transfer, whereas the other-focused configuration showed the complementary tendency toward greater transfer and lower correct resolution.

When agency became explicit:

> *If I had saved money, I would have gone on holidays.*

> *If they had saved money, they would have gone on holidays.*

the Focus distinction in resolution was substantially reduced.

This pattern makes it difficult to treat Focus as a simple pronoun contrast. Its relationship with resolution depended on how the individual was positioned within the counterfactual event structure. When the antecedent did not overtly
identify that individual as its causal agent, self- and other-focused representations were strongly differentiated. When the individual became the explicit antecedent agent, that difference largely converged.

#### Agency also changes what is represented as having gone wrong

The distinction between explicit and implicit agency becomes particularly interesting when the counterfactuals are translated back from experimental variables into the events they represent.

Compare:

> *If I had saved money, I would have gone on holidays.*

with:

> *If savings had been sufficient, I would have gone on holidays.*

Both counterfactuals concern the same individual and the same unrealised self-relevant outcome. In both cases, the individual ultimately failed to obtain the preferred outcome. What changes is how that failure is represented
within the causal structure.

In the explicit-agency version, the self is overtly represented as the agent of the antecedent action: *If I had saved money*. The counterfactual therefore directly modifies the individual's own action.

In the implicit-agency version, the self remains the experiencer of the unrealised outcome — *I would have gone on holidays* — but is no longer overtly represented as the antecedent agent. The alternative past is instead expressed through a changed circumstance: *savings had been sufficient*.

To a human interpreter, these are therefore not psychologically empty grammatical variants. The first places personal action explicitly within the causal history of the undesirable outcome; the second permits that same self-relevant outcome to be represented with less explicit attribution of personal causal responsibility.

This distinction is particularly suggestive in relation to Girotto et al.'s observation that actors tended to undo unsuccessful outcomes by altering the means or features through which their own decisions had produced them. The present experiment did not ask participants to generate counterfactual alternatives, but its Agency manipulation changed whether the self was linguistically represented as performing the antecedent action or whether the alternative was represented through a circumstance surrounding the same self-relevant outcome.

The present data cannot determine whether this difference affected perceived responsibility, controllability, self-blame, regret, psychological distance, or the alternatives participants considered available. These dimensions were
not measured. They do, however, provide theoretically meaningful candidate mechanisms for future research given that the data did show differences in the way this counterfactual considering explicit vs. implicit agency were resolved under a self-focused persepective.

The question raised by the present pattern is therefore not whether implicit-agency participants were "making excuses", but whether representing an unsuccessful self-relevant past as **something I did differently** rather than **something that happened differently** changes the processing of the counterfactual and the alternatives that subsequently become available.

The findings are therefore compatible with the broader insight motivating Girotto et al.'s actor–reader distinction: counterfactual reasoning depends not only on the objective event being represented, but also on the perspective
from which information about that event is available and organised.

This should not be interpreted as a replication of the actor–reader effect. The present participants did not differ in actual experience of the event, and the experiment did not measure which alternatives they mentally generated, which event features they considered mutable, or whether they attributed responsibility differently across conditions.

The relationship is therefore interpretive rather than evidential: the present Focus × Agency pattern shows that linguistic perspective and causal role were jointly associated with resolution behaviour, while Girotto et al. provide a useful framework for asking why the information available to a counterfactual reasoner may depend on their position within the represented event.

One particularly interesting consequence is that autobiographical participation was not necessary for a Focus distinction to emerge in the present task. Participants remained readers in both conditions, yet representing the
counterfactual from `I` rather than `They` was associated with different resolution behaviour, particularly under implicit agency.

In this limited sense, the experiment suggests a distinction between **self-focused readers and other-focused readers**: not actors and readers in Girotto et al.'s original experiential sense, but readers whose linguistic representation positioned them differently with respect to the unrealised event.

This interpretation also requires caution because the available observations were not equally distributed across the two Focus conditions. The self-focused condition contributed more observations than the other-focused condition. The apparent strength and stability of the self-focused patterns should therefore not be taken as evidence that self-focus is intrinsically more influential; the other-focused estimates were supported by fewer observations.

## 6. Food for thought: beyond the present experiment

The present study began with a relatively narrow question: whether properties of a third-conditional context could facilitate L1-Spanish learners' subsequent selection of the Past Perfect with past-subjunctive meaning.

The findings ultimately raise a broader question:

> **What if the linguistic and conceptual organisation of a learning context
> is not merely the background against which grammar is practised, but one of
> the variables through which that grammar becomes available for later use?**

The present experiment cannot establish an optimal learning sequence. Nor can its findings determine whether the observed patterns would generalise to different proficiency levels, first languages, learning environments, or productive tasks. They do, however, identify dimensions that can be experimentally manipulated rather than left incidental: **Goal structure, Focus, and Agency**.

#### From linguistic representation to psychological questions

The counterfactuals used in this experiment were linguistic stimuli, but they also represented alternative pasts involving people, actions, circumstances, goals and unrealised outcomes.

The distinction between:

> *If I had boiled the egg for longer, I would have enjoyed it better.*

and:

> *If the egg had boiled for longer, I would have enjoyed it better.*

can therefore generate questions extending beyond syntactic accessibility. Both counterfactuals concern the same self-relevant undesirable outcome, but they position the individual differently within its causal history.

Could that difference affect perceived responsibility or accountability? Could it alter controllability, self-blame, regret or psychological distance? Might an explicitly self-caused alternative make different features of an
unsuccessful event salient from one represented through changed circumstances? Could these differences, in turn, affect how competing L1 and L2 responses are explored or resolved?

The present experiment cannot answer these questions. Emotion, responsibility, regret, perceived controllability and response exploration were not measured, and none should therefore be invoked as retrospective explanations of the
observed behaviour.

They are instead **testable possibilities raised by the behavioural patterns**. Future experiments could manipulate or measure these dimensions directly and determine whether they contribute to the relationship between linguistic representation and L2 resolution.

#### From one L1 to cross-linguistic research

The same experimental architecture also invites cross-linguistic extension.

The present participants were L1-Spanish learners of English, for whom the English Past Perfect enters a particular cross-linguistic relationship with the tense and mood system of Spanish. The response patterns observed here should therefore not simply be assumed to characterise learners whose first languages organise tense, mood, counterfactuality or agency differently.

Replicating the design across different L1 populations could help distinguish patterns associated with the English target construction from patterns associated with particular L1–L2 mappings. Romance, Slavic, Germanic, Sinitic and other language backgrounds would not be expected a priori to produce particular outcomes simply because they belong to those language families; rather, their relevant linguistic systems provide new empirical comparisons through which the contribution of cross-linguistic transfer could be investigated.

#### From experimental contexts to learning trajectories

A further extension would be longitudinal.

The objective need not be to identify one configuration as universally "best" and practise it repeatedly. Instead, future research could investigate whether an empirically accessible configuration can provide an entry point from which the representational demands placed on the learner are systematically varied.

Learners might first establish reliable target use within one configuration and subsequently encounter changes in Goal, Focus and Agency, eventually requiring the same grammatical resource across increasingly varied counterfactual environments.

The relevant outcome would then extend beyond immediate accuracy:

- Does the target form remain available when the representation changes?
- Does performance persist after a delay?
- Do learners generalise to unseen lexical material?
- Can they use the form in less constrained or productive contexts?
- Do useful learning trajectories differ according to L1 or proficiency?

Large-scale digital learning environments are particularly well positioned to test such questions. Learners could be randomly exposed to systematically varied counterfactual contexts, allowing researchers to examine not only accuracy but error type, response latency, retention and generalisation. Such experiments could also directly test hypotheses about response exploration that the present offline data can only suggest.

The practical implication is therefore not that learning materials should prefer one particular combination of Goal, Focus and Agency. It is that the **conceptual organisation of practice itself may be an experimentally meaningful variable**.

For a learner, this distinction matters. The ultimate objective is not merely to recognise the appropriate form in a familiar exercise, but to have that grammatical resource available when an unfamiliar communicative situation requires it:

> *If I had considered this type of client, our turnover could have increased.*

Future research could therefore move from asking **which environment makes a target response easiest** to asking **how systematic variation across environments helps a grammatical resource become available beyond the exercise in which it was learned**.
